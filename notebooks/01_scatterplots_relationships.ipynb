# Visualizing Relationships - Scatterplots

This is the first part of this Plotly tutorial!

In this notebook, we’ll explore how to create and customize **scatter plots** using Plotly in Python. Scatter plots are one of the most fundamental forms of graphics for visualizing **relationships between two (or more) quantitative variables**. Typically, we first use them to understand correlations, clusters, and trends within a dataset.

This tutorial will guide you through:
- Creating basic scatter plots with Plotly
- Adding color, size, or shape encodings to show additional dimensions
- Enhancing interactivity with hover labels

We'll use the [Palmers Penguins](https://www.kaggle.com/datasets/satyajeetrai/palmer-penguins-dataset-for-eda) dataset found in the [data](../data) directory.

```python
# Imports
import pandas as pd
import numpy as np
import plotly.express as px
import statsmodels
```

## Creating a Basic Scatter plot

First, we'll create the most bare-bones scatter plot from our dataset. Let's read in the data.

```python
df = pd.read_csv('../data/penguins.csv')
df.head()
```

We can analyze the relationship between **bill length** and **bill depth**.

```python
fig = px.scatter(df, x="bill_length_mm", y="bill_depth_mm")
fig.show()
```

If your dataset is in the form of Numpy arrays, we can also plot the same visualization,

```python
bill_length = np.array(df["bill_length_mm"])
bill_depth = np.array(df["bill_depth_mm"])
fig = px.scatter(x = bill_length, y = bill_depth)
fig.show()
```

## Optimizing Scatter Plots

We can visually improve this plot in many ways. Here is some code to do this below.

```python
fig = px.scatter(df, x="bill_length_mm", y="bill_depth_mm", 
                 
                 # If you want to adjust the color based on data
                 color="species",
                 
                 # If you want to adjust the size based on data:
                 size='flipper_length_mm', 
                 
                 # In addition to the points defined above, if you any other information displayed on the hover label
                 hover_data=['body_mass_g'],

                 # Title setting
                 title="Palmer Penguins",

                 # The labels section below changes the names of the labels from the dataset to make it clearer. This also changes axis labels
                 labels={
                     "bill_length_mm": "Bill Length (mm)",
                     "bill_depth_mm": "Bill Depth (mm)",
                     "species": "Species",
                     "flipper_length_mm" : "Flipper Length (mm)"
                 },
                 
                 # If you wan to change the symbol based on data for CVD accessbility
                 symbol = "species"
                 )


fig.show()
```

Plotly’s `px.scatter()` makes it simple to map variables directly to axes, color, size, and more. The plot above shows several customization options, including interactive features like zooming, panning, and hover tooltips.

These interactive elements help reduce visual clutter while still conveying extra information—aligning with Tufte’s principle of maximizing the **data-ink ratio** and minimizing **chartjunk**. For example, instead of using multiple shapes or colors to show extra dimensions, we can rely on hover interactions to reveal more about each point only when needed.

When working with multivariate data, interactivity can enhance **data density** without overwhelming the viewer.

Another important design choice is **color accessibility**. As Stephen Few notes in *Practical Rules for Using Colors in Charts*, color should convey information—not just decoration. To ensure accessibility, consider using **colorblind-friendly palettes** or combining color with symbols.

Let's refine our visualization with these in mind

```python
fig = px.scatter(df, x="bill_length_mm", y="bill_depth_mm", 
                 
                 # If you want to adjust the color based on data
                 color="flipper_length_mm",
                 
                 # If you want to adjust the size based on data:
                 size='body_mass_g', 
                 
                 # In addition to the points defined above, if you any other information displayed on the hover label
                 hover_data=['body_mass_g'],

                 # Title setting
                 title="Palmer Penguins",

                 # The labels section below changes the names of the labels from the dataset to make it clearer. This also changes axis labels
                 labels={
                     "bill_length_mm": "Bill Length (mm)",
                     "bill_depth_mm": "Bill Depth (mm)",
                     "species": "Species",
                     "flipper_length_mm" : "Flipper Length (mm)",
                     "body_mass_g" : "Body Mass (g)"
                 },
                 
                 # If you wan to change the symbol based on data for CVD accessbility
                 symbol = "species",

                # Adjust the size to make the plot look more clean
                 width=1000, height=800

                 )

# To make our figure look cleaner and legends don't overlap
fig.update_layout(
    margin=dict(l=50, r=50, t=80, b=50),  # Add margins
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)  # Move legend to top
)


fig.show()
```

Analyzing our plot, we can already visually see that
- Adelie penguins are more concentrated on the top-left of the graph (smaller bill length, larger bill depth)
- Chinstrap penguins are concentrated on the top-right of the graph (larger bill length, larger bill depth)
- Gentoo penguins are concentrated on the bottom-right of the graph (larger bill length, smaller bill depth)
- As bill depth increases, the flipper length decreases, although we cannot deduce if it is linear
- There is not a clear linear relationship between bill depth and bill length.

Although these can be seen from our visualization, as Albert Cairo highlights in Chapter 9 of *The Trufthful Art: Data, Charts and Maps for Communication*, it is important to evaluate other metrics before drawing strong conclusions.

## Additional Modifications

There are other aspects of a scatterplot that weren't necessarily highlighted in our final visualization, but are worth noting.

### Linear Best fit Lines

We can use these to gain insight into the "strength" of the linear relationship between two variables.

```python
fig = px.scatter(df, x="bill_depth_mm", y = "flipper_length_mm", 
                 
                 # In addition to the points defined above, if you any other information displayed on the hover label
                 hover_data=['body_mass_g'],

                 # Title setting
                 title="Palmer Penguins",

                 # The labels section below changes the names of the labels from the dataset to make it clearer. This also changes axis labels
                 labels={
                     "bill_length_mm": "Bill Length (mm)",
                     "bill_depth_mm": "Bill Depth (mm)",
                     "species": "Species",
                     "flipper_length_mm" : "Flipper Length (mm)",
                     "body_mass_g" : "Body Mass (g)"
                 },

                # Adjust the size to make the plot look more clean
                 width=1000, height=800,

                # Adds a trend line based on Ordinary Least Squares
                trendline = 'ols'

                 )

fig.show()
```

### Facetting

We can use this to analyze certain sections of the dataset.



```python
fig = px.scatter(df, x="bill_depth_mm", y = "flipper_length_mm", 
                 
                 # In addition to the points defined above, if you any other information displayed on the hover label
                 hover_data=['body_mass_g'],

                 # Adjusts color according to this variable in the data
                 color = 'species',

                 # Title setting
                 title="Palmer Penguins",

                 # The labels section below changes the names of the labels from the dataset to make it clearer. This also changes axis labels
                 labels={
                     "bill_length_mm": "Bill Length (mm)",
                     "bill_depth_mm": "Bill Depth (mm)",
                     "species": "Species",
                     "flipper_length_mm" : "Flipper Length (mm)",
                     "body_mass_g" : "Body Mass (g)"
                 },

                 # Adjust the size to make the plot look more clean
                 width=1000, height=800,

                 # Splits the columns according to this variable
                 facet_col="species",

                 # Splits the rows according to this variable
                 # facet_row = "species"

                 trendline = 'ols'

                 )

fig.show()
```

### Marginal Distribution Plots

We can use these plots to gain insight into the distribution of certain variables in addition to seeing their relationship from a scatter plot.

```python
fig = px.scatter(df, x="bill_depth_mm", y = "flipper_length_mm", 
                 
                 # In addition to the points defined above, if you any other information displayed on the hover label
                 hover_data=['body_mass_g'],

                 # Title setting
                 title="Palmer Penguins",

                 # The labels section below changes the names of the labels from the dataset to make it clearer. This also changes axis labels
                 labels={
                     "bill_length_mm": "Bill Length (mm)",
                     "bill_depth_mm": "Bill Depth (mm)",
                     "species": "Species",
                     "flipper_length_mm" : "Flipper Length (mm)",
                     "body_mass_g" : "Body Mass (g)"
                 },

                # Adjust the size to make the plot look more clean
                 width=1000, height=800,
                
                # Set what kind of a distribution plot you want to make for the variables on the x and y axis
                marginal_x="histogram", marginal_y="histogram"

                 )

fig.show()
```
