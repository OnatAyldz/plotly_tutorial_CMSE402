# Visualizing Distributions and Statistical Variation: Histograms, Box Plots, and Violin Plots

This is the fourth notebook in the Plotly tutorial series.

Understanding the distribution of data is fundamental to analysis. Distribution plots help us analyze the spread of the data, outliers in the data, and symmetry and skewedness of the data.

In this notebook, we’ll explore several common and powerful techniques for visualizing distributions using Plotly:

- **Histograms**: useful for showing the frequency of values within intervals and identifying the shape of the distribution.
- **Box plots**: provides key statistics like median, quartiles, and potential outliers; especially useful for comparisons across groups.
- **Violin plots**: combine box plots with kernel density estimates, offering a more in-depth picture of the shape of the distribution

Specifically, this notebook covers:
- Creating basic histograms, box plots and violin plots
- Advantages of each form of visualization for different tasks
- Customizing layout, colors, and interactivity
- Best practices for visualizations to analyze distributions using Plotly

For these visualizations, we will be using the [Palmers Penguins](https://www.kaggle.com/datasets/satyajeetrai/palmer-penguins-dataset-for-eda) dataset, found in the [data directory](../data)

```python
# Imports
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import statsmodels
```

## Histograms

Histograms are one of the most fundamental tools for visualizing the distribution of a single continuous variable. They divide the data into bins and count how many values fall into each bin, helping us see patterns such as skewness, modality (e.g., unimodal, bimodal), and spread. Plotly allows for flexible customization, including stacking or overlaying multiple distributions and controlling bin sizes for more precise interpretations.

```python
# Load in data
df = pd.read_csv('../data/penguins.csv')
df.dropna(inplace = True)
df.head()
```

```python
fig = px.histogram(df, x="flipper_length_mm", nbins=20,
                   title="Distribution of Flipper Length")
fig.show()
```

So we can see that there are two peaks in this distribution. A common flipper length being around 190-200 and the second most common being around 210-220. We can gain more insight into our data and distribution of flipper lengths by adjust the visualization in the following way:

```python
fig = px.histogram(df, x="flipper_length_mm", nbins=40, color = 'species', title="Distribution of Flipper Length by Penguin Species",
                   
                   # Adding numbers allows for better quantitative comparison since bars are stacked
                   text_auto = True,

                   labels = {"count": "Count",
                             "species": "Species",
                             "flipper_length_mm": "Flipper Length(mm)"},

                   height = 800, width = 1000
                   )

# Introduce a gap between bars
fig.update_layout(bargap = 0.1)
fig.show()
```

We can visualize this in a different way by overlaying these distributions as follows:

```python
import plotly.graph_objects as go

fig = go.Figure()
fig.add_trace(go.Histogram(x=df[df['species'] == 'Gentoo']['flipper_length_mm'], name='Gentoo'))
fig.add_trace(go.Histogram(x=df[df['species'] == 'Adelie']['flipper_length_mm'], name='Adelie'))
fig.add_trace(go.Histogram(x=df[df['species'] == 'Chinstrap']['flipper_length_mm'], name='Chinstrap'))

# Overlay both histograms
fig.update_layout(
    barmode='overlay',
    width=800,
    height=500,
    title = "Distribution of Flipper Length by Penguin Species"
)

# Reduce opacity to see both histograms
fig.update_traces(opacity=0.75)
fig.show()
```

## Box Plots

Box plots (or box-and-whisker plots) provide a summary of a distribution’s mean, variance, and potential outliers. They are especially useful for comparing multiple groups or categories side-by-side. In line with Edward Tufte’s principles of maximizing data density and reducing chartjunk, box plots convey a great deal of statistical information using minimal visual elements, although in Chapter 2 of *The Visual Display of Quantitative Information*, he discusses the potential to reduce even this plot's data-ink-ratio further.

We can create a very simple boxplot from our data as shown:

```python
fig = px.box(df, x="species", y="body_mass_g", color="species",
             title="Body Mass Distribution by Species",)
fig.show()
```

We can see the distributions of the body masses of the penguins. Not only do we see them relative to each other, but we can quantitatively see the mean, variance, range and outliers of our data for each category. Plotly makes this extra simple since you can just enter a column name for the `x` argument in the `px.box()` function and it will create a boxplot for each unique value in the column. Additionally, when you hover over one of the boxplots, it will show you the statistical quantities with text, making it extra powerful for drawing statistical insights.4

You can also include all of the data with your boxplots as shown below:

```python
fig = px.box(df, x="species", y="body_mass_g", color="species",
             title="Body Mass Distribution by Species", points = 'all',
             width = 800, height = 600)
fig.show()
```

You can also add notches to make the boxplots appear more visually pleasing, following Albert Cairo's guidelines for the importance of aesthetics in visualizations in *The Truthful Art: Data, Charts and Maps for Communication*.

```python
fig = px.box(df, x="species", y="body_mass_g", color="species",
             title="Body Mass Distribution by Species", points = 'all',
             width = 800, height = 600, notched = True)
fig.show()
```

## Violin Plots

Violin plots combine the summary statistics of box plots with a mirrored kernel density plot, offering a fuller picture of a distribution’s shape. This is particularly useful when two groups have similar medians and interquartile ranges but differ in how their data is distributed. Plotly’s implementation enables side-by-side comparison and interactive exploration of data nuances, making violin plots a powerful extension of basic statistical visualizations.

```python
fig = px.violin(df, x="species", y="flipper_length_mm", color="species",
                title="Flipper Length Distribution by Sex")
fig.show()
```

We can further customize these violin plots by adding boxplots within them, showing the points, and overall giving it a better layout.

```python
fig = px.violin(df, x="species", y="flipper_length_mm", color="species", box=True, points="all",
                title="Flipper Length Distribution by Species",
                labels = {'flipper_length_mm' : 'Flipper Length (mm)',
                         'species': 'Species'},
                width = 1000, height = 600
)
fig.show()
```

One benefit of violin plots is the visual affect it has on the viewer. Boxplots are nice to show statistical information, but it might not give the viewer an idea of the distribution immediately. The advantage of the violin plot is that the shape directly communicates to the viewer what the distribution looks like, giving them an idea before they even analyze the quantitative information. So it leverages the preattentive features discussed in Christopher Healey's *Perception in Visualization*.

A shortcoming of typical violin plots is that they don't show much quantitative information. However, not only does Plotly's boxplot overlay allow for this to be demonstrated, but its hover label interactivity also allows us to see all of the statistical information that we saw on the boxplots, giving us the best of both worlds.

We can even further improve our violin plots like so, to highlight more dimensions of the data:

```python
fig = px.violin(df, x="species", y="flipper_length_mm", color="sex", box=True, points="all",
                title="Flipper Length Distribution by Species",
                labels = {'flipper_length_mm' : 'Flipper Length (mm)',
                         'species': 'Species',
                         'sex': 'Sex'},
                width = 1100, height = 600
)
fig.show()
```

## Final Remarks

Visualizing distributions and statistical variation helps us gain a deeper understanding of the underlying structure of our data:

- **Histograms** reveal the overall shape, central tendencies, and spread.
- **Box plots** offer concise visual summaries of key statistics and help identify outliers.
- **Violin plots** combine statistical summaries with distributional shape, giving a more comprehensive view of group differences.

By using Plotly’s interactive features, we can explore these aspects dynamically by hovering over data points or zooming into areas of interest. These visualizations are foundational for statistical analysis and data exploration, especially in the early data analysis phase.
