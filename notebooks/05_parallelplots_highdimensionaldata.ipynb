# Visualizing Multivariate and High-Dimensional Data with Plotly

This is the fifth notebook in the Plotly tutorial series.

Real-world datasets often contain many variables, and understanding how they interact can be complex. Multivariate visualizations aim to show relationships across more than two or three dimensions in a way that is still interpretable and insightful.

In this notebook, we’ll explore techniques for visualizing high-dimensional data using Plotly:

- **Scatter plots with color, size, and symbol encodings**: ideal for representing multiple dimensions in a 2D space.
- **Parallel coordinates plots**: useful for visualizing patterns across multiple numerical variables and identifying correlations or clusters.

Specifically, this notebook covers:
- Creating scatter plots with more encodings (bubble charts) and parallel coordinates plots
- Advantages of each form of visualization for different tasks
- Customizing layout, colors, and interactivity
- Best practices for visualizations to analyze multivariate data using Plotly

For these visualizations, we will be using the [Palmers Penguins](https://www.kaggle.com/datasets/satyajeetrai/palmer-penguins-dataset-for-eda) dataset, found in the [data directory](../data)


```python
# Imports
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import statsmodels
```

## Bubble charts

Bubble charts are an extension of scatter plots that incorporate a third numeric variable through the size of the data points. They are ideal for simultaneously visualizing relationships between two continuous variables while encoding an additional dimension through size. This allows for richer insight into multivariate data in a two-dimensional space.

```python
# Load in data
df = pd.read_csv('../data/penguins.csv')
df.dropna(inplace = True)
df.head()
```

```python
fig = px.scatter(df, x="bill_length_mm", y="bill_depth_mm", 
                 
                 # If you want to adjust the color based on data
                 color="species",
                 
                 # If you want to adjust the size based on data:
                 size='flipper_length_mm', 
                 
                 # In addition to the points defined above, if you any other information displayed on the hover label
                 hover_data=['body_mass_g'],

                 # Title setting
                 title="Palmer Penguins",

                 # The labels section below changes the names of the labels from the dataset to make it clearer. This also changes axis labels
                 labels={
                     "bill_length_mm": "Bill Length (mm)",
                     "bill_depth_mm": "Bill Depth (mm)",
                     "species": "Species",
                     "flipper_length_mm" : "Flipper Length (mm)"
                 },
                 
                 # If you wan to change the symbol based on data for CVD accessbility
                 symbol = "species"
                 )


fig.show()
```

Plotly’s `px.scatter()` makes it simple to map variables directly to axes, color, size, and more. The plot above shows several customization options, including interactive features like zooming, panning, and hover tooltips.

These interactive elements help reduce visual clutter while still conveying extra information—aligning with Tufte’s principle of maximizing the **data-ink ratio** and minimizing **chartjunk**. For example, instead of using multiple shapes or colors to show extra dimensions, we can rely on hover interactions to reveal more about each point only when needed.

When working with multivariate data, interactivity can enhance **data density** without overwhelming the viewer.

Another important design choice is **color accessibility**. As Stephen Few notes in *Practical Rules for Using Colors in Charts*, color should convey information—not just decoration. To ensure accessibility, consider using **colorblind-friendly palettes** or combining color with symbols.

Let's refine our visualization with these in mind

```python
fig = px.scatter(df, x="bill_length_mm", y="bill_depth_mm", 
                 
                 # If you want to adjust the color based on data
                 color="flipper_length_mm",
                 
                 # If you want to adjust the size based on data:
                 size='body_mass_g', 
                 
                 # In addition to the points defined above, if you any other information displayed on the hover label
                 hover_data=['body_mass_g'],

                 # Title setting
                 title="Palmer Penguins",

                 # The labels section below changes the names of the labels from the dataset to make it clearer. This also changes axis labels
                 labels={
                     "bill_length_mm": "Bill Length (mm)",
                     "bill_depth_mm": "Bill Depth (mm)",
                     "species": "Species",
                     "flipper_length_mm" : "Flipper Length (mm)",
                     "body_mass_g" : "Body Mass (g)"
                 },
                 
                 # If you wan to change the symbol based on data for CVD accessbility
                 symbol = "species",

                # Adjust the size to make the plot look more clean
                 width=1000, height=800

                 )

# To make our figure look cleaner and legends don't overlap
fig.update_layout(
    margin=dict(l=50, r=50, t=80, b=50),  # Add margins
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)  # Move legend to top
)


fig.show()
```

Analyzing our plot, we can already visually see that
- Adelie penguins are more concentrated on the top-left of the graph (smaller bill length, larger bill depth)
- Chinstrap penguins are concentrated on the top-right of the graph (larger bill length, larger bill depth)
- Gentoo penguins are concentrated on the bottom-right of the graph (larger bill length, smaller bill depth)
- As bill depth increases, the flipper length decreases, although we cannot deduce if it is linear
- There is not a clear linear relationship between bill depth and bill length.

Although these can be seen from our visualization, as Albert Cairo highlights in Chapter 9 of *The Trufthful Art: Data, Charts and Maps for Communication*, it is important to evaluate other metrics before drawing strong conclusions.

The strength of the bubble chart is that you can visually encode many different aspects of the same dataset. For example, the following bubble chart is a different analysis, using the same data, which allows us to see the clusters in the sexes in addition to the species encoded through symbols.

```python
fig = px.scatter(df, x="bill_length_mm", y="bill_depth_mm", 
                 
                 # If you want to adjust the color based on data
                 color="sex",
                 
                 # If you want to adjust the size based on data:
                 size='flipper_length_mm', 
                 
                 # In addition to the points defined above, if you any other information displayed on the hover label
                 hover_data=['body_mass_g'],

                 # Title setting
                 title="Palmer Penguins",

                 # The labels section below changes the names of the labels from the dataset to make it clearer. This also changes axis labels
                 labels={
                     "bill_length_mm": "Bill Length (mm)",
                     "bill_depth_mm": "Bill Depth (mm)",
                     "species": "Species",
                     "flipper_length_mm" : "Flipper Length (mm)",
                     "body_mass_g" : "Body Mass (g)"
                 },
                 
                 # If you wan to change the symbol based on data for CVD accessbility
                 symbol = "species",

                # Adjust the size to make the plot look more clean
                 width=1000, height=800

                 )

# To make our figure look cleaner and legends don't overlap
fig.update_layout(
    margin=dict(l=50, r=50, t=80, b=50),  # Add margins
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)  # Move legend to top
)


fig.show()
```

## Parallel Coordinate Plots

Parallel coordinate plots are powerful tools for visualizing high-dimensional data. Each variable is represented as a vertical axis, and each data point is drawn as a line connecting its values across these axes. This technique helps reveal patterns, correlations, and clusters across many variables at once.

```python
# You must create numerical encodings for categorical variables otherwise the function won't work
df['species_id'] = pd.Categorical(df['species']).codes
df['island_id'] = pd.Categorical(df['island']).codes

fig = px.parallel_coordinates(
    df,
    color="species_id",
    dimensions=["island_id", "flipper_length_mm", "bill_depth_mm", "bill_length_mm", "body_mass_g"],
    color_continuous_scale=px.colors.qualitative.Plotly,
    title="Parallel Coordinates Plot of Palmers Penguins"
)

fig.show()
```

It is important to note that we had to numerically encode some of our categorical variables here for the `px.parallel_coordinates()` function to work, but this doesn't give us much information. We can modify this by updating the visualization after the function has been run as follows:

```python
# You must create numerical encodings for categorical variables otherwise the function won't work
df['species_id'] = pd.Categorical(df['species']).codes
df['island_id'] = pd.Categorical(df['island']).codes

fig = px.parallel_coordinates(
    df,
    color="species_id",
    dimensions=["island_id", "flipper_length_mm", "bill_depth_mm", "bill_length_mm", "body_mass_g"],
    color_continuous_scale=px.colors.sequential.Viridis,
    title="Parallel Coordinates Plot of Palmers Penguins"
)

# Create mapping dictionaries
island_labels = {code: island for code, island in enumerate(df['island'].unique())}
species_labels = {code: species for code, species in enumerate(df['species'].unique())}

# Update the island axis with proper labels
fig.update_traces(
    dimensions=[
        dict(
            label="Island",
            tickvals=list(island_labels.keys()),
            ticktext=list(island_labels.values())
        ),
        dict(label="Flipper Length (mm)"),
        dict(label="Bill Depth (mm)"),
        dict(label="Bill Length (mm)"),
        dict(label="Body Mass (g)")
    ]
)

# Update the color axis with proper species labels
fig.update_layout(
    coloraxis_colorbar=dict(
        title="Species",
        tickvals=list(species_labels.keys()),
        ticktext=list(species_labels.values())
    )
)
```

This visualization allows us to clearly see the different clusters across multiple variables in the Palmers penguins dataset, by following the colors of each line. Another very powerful aspect of Plotly is that it automatically normalizes the data for each category, so we don't have to manually do it the same way we did for the radar chart. In essence, the Parallel Coordinate plot could be considered an improvement to the radar chart, showing it linearly and more clearly. From our visualization we can conclude:
- What island has each kind of penguin, dream has only Adelie, Biscoe has Adelie and Gentoo and Torgersen has Adelie and Chinstrap
- Chinstrap penguins distinctly have higher flippper lengths while the other two species are relatively close
- Chinstrap penguins distinctly have lower bill depths while the other two are relatively close
- Adelie penguins have distinctly low bill lengths while the other two species are relatively close
- Chinstrap penguins have a distinctly higher body mass while the other two species are relatively close.

## Final Remarks

Visualizing multivariate and high-dimensional data presents a unique challenge: how do we represent many variables in a way that remains clear and insightful? In this notebook, we explored several effective strategies using Plotly:

- Bubble charts expand the basic scatter plot with additional dimensions via size, color and symols, helping us encode up to five variables at once.
- Parallel coordinate plots allowed us to visualize entire records across many dimensions, revealing patterns, clusters, and outliers in high-dimensional data.

Each of these visualizations serves a different but complementary purpose in exploring complex, multivariate datasets. They can help show patterns that would not be visible in lower-dimensional visualizations or purely statistical summaries. And by using Plotly’s interactive features, we can explore these aspects dynamically by hovering over data points or zooming into areas of interest. These visualizations are foundational for statistical analysis and data exploration, especially in the early data analysis phase.
